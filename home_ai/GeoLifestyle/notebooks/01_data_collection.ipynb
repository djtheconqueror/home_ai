```python
from openmeteo_requests import Client
import numpy as np

client = Client()

temps = []
for c in df["city"]:
    city_name = c.split(",")[0]
    try:
        resp = client.free(text=city_name)
        temps.append(resp.current.temperature_2m)
    except:
        temps.append(None)

df["avg_temp"] = temps
df
```

```python
from census import Census
from us import states
import pandas as pd

CENSUS_KEY = "7e2aa0723914b4b33da982ed7ef78f11074bd7df"  # replace but do NOT commit to git
c = Census(CENSUS_KEY)

diversity_data = []

for cty in df["city"]:
    name, state = cty.split(", ")
    try:
        pop = c.acs5.table("B03002", state=states.lookup(state).fips, county=None)
    except:
        diversity_data.append(None)
        continue

    # Extract race counts
    record = pop[0]
    # Racial categories used in diversity index
    groups = [
        record["B03002_003E"],  # White alone
        record["B03002_004E"],  # Black/African American alone
        record["B03002_006E"],  # Asian alone
        record["B03002_007E"],  # Native American/Alaska Native alone
        record["B03002_012E"],  # Two or more races
        record["B03002_005E"] + record["B03002_008E"] + record["B03002_009E"]  # Hispanic aggregated
    ]
    
    total = sum(groups)
    p = [x/total for x in groups if x > 0]

    # Shannon Diversity Index
    import numpy as np
    diversity = -sum([x*np.log(x) for x in p])
    diversity_data.append(diversity)

df["diversity_index"] = diversity_data
df
```
