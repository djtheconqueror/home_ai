```python
import pandas as pd

data = {
    "city": ["Charlotte", "Raleigh", "Atlanta"],
    "state": ["NC", "NC", "GA"],
    "rent": [1700, 1650, 1800],
    "temperature": [68, 66, 72],
    "diversity_score": [0.81, 0.79, 0.83]
}

df = pd.DataFrame(data)
df.to_csv("cities.csv", index=False)

df
```

```python
def score_city(row, 
               pref_region=['East','South'],
               pref_min_rent=1500,
               pref_max_rent=2500,
               pref_climate=['Mild','Warm'],
               pref_diversity='High',
               pref_lifestyle='Balanced'):

    score = 0

    # Region
    if row['region'] in pref_region:
        score += 20

    # Rent
    if pref_min_rent <= row['rent'] <= pref_max_rent:
        score += 20

    # Climate
    if row['climate'] in pref_climate:
        score += 20

    # Diversity
    if row['diversity'] == pref_diversity:
        score += 20

    # Lifestyle
    if row['lifestyle'] == pref_lifestyle:
        score += 20

    return score
```

```python
def score_city(row, 
               pref_min_rent, pref_max_rent, 
               pref_min_temp, pref_max_temp,
               pref_diversity):
    
    score = 0
    
    # Rent score
    if pref_min_rent <= row['rent'] <= pref_max_rent:
        score += 30

    # Temperature score
    if pref_min_temp <= row['temperature'] <= pref_max_temp:
        score += 30

    # Diversity score (0–100)
    score += (row['diversity_score'] / 100) * 20

    return score
```

```python
df['score'] = df.apply(
    lambda row: score_city(
        row,
        pref_min_rent=1500,
        pref_max_rent=2500,
        pref_min_temp=60,
        pref_max_temp=85,
        pref_diversity=60
    ), 
    axis=1
)

df.sort_values('score', ascending=False).head(10)
```
